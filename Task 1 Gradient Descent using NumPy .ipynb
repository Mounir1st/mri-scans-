```python
import numpy as np

# Initial values
r = 3.0
s = 1.5
t = 4.0

# Hyperparameters
learning_rate = 1e-6
max_iters = 200000
tolerance = 1e-12
```

```python
# Function y
def y_function(r, s, t):
    return ((r**2 - 81)**2 +
            (24*s - 6)**2 +
            (20*s - 5)**2 +
            ((5*t**2)/t - 50)**2)

# Error function E = y^2
def error_function(r, s, t):
    y = y_function(r, s, t)
    return y**2
```

```python
# Gradients of y
def dy_dr(r):
    return 4 * r * (r**2 - 81)

def dy_ds(s):
    return 48 * (24*s - 6) + 40 * (20*s - 5)

def dy_dt(t):
    # Since (5t^2)/t = 5t, this becomes derivative of (5t - 50)^2
    return 10 * (5*t - 50)

```

```python
for i in range(max_iters):
    y = y_function(r, s, t)

    # gradients of y
    grad_r = dy_dr(r)
    grad_s = dy_ds(s)
    grad_t = dy_dt(t)

    # update
    r = r - learning_rate * grad_r
    s = s - learning_rate * grad_s
    t = t - learning_rate * grad_t

    if i % 5000 == 0:
        print(f"Iteration {i}: r={r:.6f}, s={s:.6f}, t={t:.6f}, y={y:.12f}")

    if y < tolerance:
        print(f"\nConverged after {i+1} iterations")
        break
```

```python
tolerance = 1e-6
final_y = y_function(r, s, t)
final_E = final_y**2

print("Final values after gradient descent:")
print(f"r = {r:.6f}")
print(f"s = {s:.6f}")
print(f"t = {t:.6f}")
print(f"y = {final_y:.12f}")
print(f"E = {final_E:.12f}")
```
