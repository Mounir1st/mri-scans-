```python
import torch

# Q1
x = torch.tensor(1.0, requires_grad=True)
c = torch.tensor(1.0)   # constant

s = x**2 + 1
t = 10*s + 14*c*s
y = 10*t**2 + 25*t + (4*c**2)/c

y.backward()

print("Q1")
print("y =", y.item())
print("dy/dx =", x.grad.item())
```

```python
# Q2
x = torch.tensor(1.0, requires_grad=True)
c = torch.tensor(1.0)   # constant

t = torch.exp(x) + c
z = t**0.5 + 2*t
y = 5*z + 1

y.backward()

print("Q2")
print("y =", y.item())
print("dy/dx =", x.grad.item())
```

```python
# Q3
x = torch.tensor(1.0, requires_grad=True)
c = torch.tensor(1.0)   # constant

y = 5*torch.log(x) + 25*x**2 + 2*c**2

y.backward()

print("Q3")
print("y =", y.item())
print("dy/dx =", x.grad.item())
```
